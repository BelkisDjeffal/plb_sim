# PLB policy plots

Plots for class-blind baselines and the N-class PLB scheduler.

```python
from pathlib import Path
import json

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd()
SCENARIO = "no_fault_3_classes"
REP = "rep_01"
POLICIES = ["round_robin", "least_loaded", "plb_nclass"]
THETA = 10
BIN_SIZE_S = 5

CLASS_ORDER = ["enterprise", "premium", "freemium"]
CLASS_COLORS = {
    "enterprise": "#CC79A7",
    "premium": "#E69F00",
    "freemium": "#0D7D87",
}
POLICY_COLORS = {
    "round_robin": "#5F6B73",
    "least_loaded": "#C9CFD6",
    "plb_nclass": "#111111",
}

OUT = ROOT / "notebook_plots" / SCENARIO
OUT.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({
    "figure.figsize": (8, 4.8),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "legend.frameon": False,
})
```

## Load files

```python
def policy_dir(policy):
    return ROOT / "outputs" / SCENARIO / REP / policy

def read_csv_if_exists(path):
    return pd.read_csv(path) if path.exists() and path.stat().st_size > 0 else pd.DataFrame()

def read_decisions(policy):
    df = read_csv_if_exists(policy_dir(policy) / "scheduler_decisions.csv")
    if not df.empty:
        df["policy"] = policy
    return df

decisions = pd.concat([read_decisions(p) for p in POLICIES], ignore_index=True)
decisions.head()
```

## Arrivals per time bin

This describes the generated workload, not a scheduler effect.

```python
summary_path = ROOT / "workloads" / f"{SCENARIO}_summary.csv"
workload = pd.read_csv(summary_path)
workload["bin"] = (workload["subtime_s"] // BIN_SIZE_S * BIN_SIZE_S).astype(int)
arrivals = (
    workload.groupby(["bin", "class"])
    .size()
    .unstack(fill_value=0)
    .reindex(columns=CLASS_ORDER, fill_value=0)
)

ax = arrivals.plot(
    kind="bar",
    stacked=True,
    width=0.9,
    color=[CLASS_COLORS[c] for c in arrivals.columns],
)
ax.set_title(f"Arrivals per {BIN_SIZE_S}s bin")
ax.set_xlabel("Simulation time bin (s)")
ax.set_ylabel("Arriving sessions")
ax.legend(title="Class")
plt.tight_layout()
plt.savefig(OUT / "arrivals_per_bin_by_class.png", dpi=220)
plt.show()
```

## Started sessions per replica

```python
for policy in POLICIES:
    df = decisions[(decisions["policy"] == policy) & (decisions["event"] == "start")].copy()
    if df.empty:
        continue
    df["replica"] = df["replica"].astype(int)
    table = (
        df.groupby(["replica", "class"])
        .size()
        .unstack(fill_value=0)
        .reindex(columns=CLASS_ORDER, fill_value=0)
        .sort_index()
    )
    ax = table.plot(
        kind="bar",
        stacked=True,
        color=[CLASS_COLORS[c] for c in table.columns],
        width=0.82,
    )
    ax.set_title(f"Started sessions per replica, {policy}")
    ax.set_xlabel("Logical replica")
    ax.set_ylabel("Started sessions")
    ax.legend(title="Class")
    plt.tight_layout()
    plt.savefig(OUT / f"{policy}_started_sessions_per_replica.png", dpi=220)
    plt.show()
```

## Saturation exposure by class

Exposure is the fraction of class session-seconds spent on replicas with total active sessions greater than or equal to theta.

```python
frames = []
for policy in POLICIES:
    df = read_csv_if_exists(policy_dir(policy) / "saturation_exposure_by_class.csv")
    if not df.empty:
        df["policy"] = policy
        frames.append(df)

exposure = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()
if not exposure.empty:
    exposure["class"] = pd.Categorical(exposure["class"], categories=CLASS_ORDER, ordered=True)
    pivot = exposure.pivot(index="class", columns="policy", values="saturation_exposure_pct").reindex(CLASS_ORDER)
    ax = pivot.plot(
        kind="bar",
        width=0.82,
        color=[POLICY_COLORS.get(c, None) for c in pivot.columns],
    )
    ax.set_title("Class exposure to saturated replicas")
    ax.set_xlabel("Class")
    ax.set_ylabel("Saturated session-seconds (%)")
    ax.legend(title="Policy")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(OUT / "comparison_saturation_exposure_by_class.png", dpi=220)
    plt.show()

exposure
```

## Peak active sessions per replica

```python
frames = []
for policy in POLICIES:
    df = read_csv_if_exists(policy_dir(policy) / "replica_load_summary.csv")
    if not df.empty:
        df["policy"] = policy
        frames.append(df)
replica_summary = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()

for policy in POLICIES:
    df = replica_summary[replica_summary["policy"] == policy].copy()
    if df.empty:
        continue
    fig, ax = plt.subplots()
    ax.bar(df["replica"], df["max_active"], color=POLICY_COLORS.get(policy, None))
    ax.axhline(THETA, linestyle="--", linewidth=1.2, label=f"theta={THETA}")
    ax.set_title(f"Peak active sessions per replica, {policy}")
    ax.set_xlabel("Logical replica")
    ax.set_ylabel("Peak active sessions")
    ax.legend()
    plt.tight_layout()
    plt.savefig(OUT / f"{policy}_peak_active_sessions_per_replica.png", dpi=220)
    plt.show()
```

## Imbalance summary

```python
frames = []
for policy in POLICIES:
    df = read_csv_if_exists(policy_dir(policy) / "imbalance_summary.csv")
    if not df.empty:
        df["policy"] = policy
        frames.append(df)
imbalance = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()

if not imbalance.empty:
    melted = imbalance.melt(
        id_vars="policy",
        value_vars=["mean_imbalance", "p95_imbalance", "max_imbalance"],
        var_name="metric",
        value_name="imbalance",
    )
    pivot = melted.pivot(index="metric", columns="policy", values="imbalance")
    ax = pivot.plot(kind="bar", width=0.82, color=[POLICY_COLORS.get(c, None) for c in pivot.columns])
    ax.set_title("Replica load imbalance summary")
    ax.set_xlabel("Metric")
    ax.set_ylabel("Max replica load - min replica load")
    ax.legend(title="Policy")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(OUT / "comparison_imbalance_summary.png", dpi=220)
    plt.show()

imbalance
```

## Pool sizes over time

This is mostly useful for `plb_nclass`.

```python
for policy in POLICIES:
    pool_ts = read_csv_if_exists(policy_dir(policy) / "pool_size_timeseries.csv")
    if pool_ts.empty:
        continue
    pool_cols = [c for c in [*CLASS_ORDER, "mixed"] if c in pool_ts.columns]
    if not pool_cols:
        continue
    fig, ax = plt.subplots()
    for col in pool_cols:
        color = CLASS_COLORS.get(col, POLICY_COLORS.get(policy, None))
        ax.step(pool_ts["time_s"], pool_ts[col], where="post", linewidth=2, label=col, color=color)
    ax.set_title(f"Pool sizes over time, {policy}")
    ax.set_xlabel("Simulation time (s)")
    ax.set_ylabel("Replicas held")
    ax.legend(title="Pool")
    plt.tight_layout()
    plt.savefig(OUT / f"{policy}_pool_sizes_over_time.png", dpi=220)
    plt.show()
```

## Borrowing and return events

```python
frames = []
for policy in POLICIES:
    df = read_csv_if_exists(policy_dir(policy) / "borrow_events.csv")
    if not df.empty:
        df["policy"] = policy
        frames.append(df)
events = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()

if not events.empty:
    counts = events.groupby(["policy", "action"]).size().reset_index(name="count")
    pivot = counts.pivot(index="action", columns="policy", values="count").fillna(0)
    ax = pivot.plot(kind="bar", width=0.82, color=[POLICY_COLORS.get(c, None) for c in pivot.columns])
    ax.set_title("Borrowing and return events")
    ax.set_xlabel("Action")
    ax.set_ylabel("Event count")
    ax.legend(title="Policy")
    plt.xticks(rotation=25, ha="right")
    plt.tight_layout()
    plt.savefig(OUT / "borrowing_return_events.png", dpi=220)
    plt.show()

events.head(20)
```

## Connection lifespans per replica

This plot shows the lifetime of each simulated worker/session on the replica selected by a policy.
Each horizontal segment is one job/session. The color is the priority class.


```python
# Connection lifespans per replica
import pandas as pd
from pathlib import Path

try:
    import plotly.express as px
except ImportError as exc:
    raise ImportError("Install Plotly with: python -m pip install plotly") from exc

LIFESPAN_POLICY = "plb_nclass"
MAX_WORKERS_PER_REPLICA = None

run_dir = policy_dir(LIFESPAN_POLICY)
jobs_path = run_dir / "out__jobs.csv"
decisions_path = run_dir / "scheduler_decisions.csv"

def read_csv_auto(path):
    return pd.read_csv(path, sep=None, engine="python")

def canonical_job_id(value):
    value = str(value)
    if "!" in value:
        value = value.split("!", 1)[1]
    return value

jobs = read_csv_auto(jobs_path)
decisions = read_csv_auto(decisions_path)

jobs.columns = [str(c).strip() for c in jobs.columns]
decisions.columns = [str(c).strip() for c in decisions.columns]

start_events = decisions[decisions["event"].astype(str).eq("start")].copy()
start_events = start_events[["job_id", "class", "replica"]].drop_duplicates("job_id")
start_events["job_key"] = start_events["job_id"].map(canonical_job_id)

job_id_col = "job_id"
start_col = "starting_time" if "starting_time" in jobs.columns else "start_time"
finish_col = "finish_time" if "finish_time" in jobs.columns else None

if finish_col is None:
    duration_col = "execution_time" if "execution_time" in jobs.columns else "runtime"
    jobs["finish_time"] = jobs[start_col].astype(float) + jobs[duration_col].astype(float)
    finish_col = "finish_time"

lifespan_df = jobs[[job_id_col, start_col, finish_col]].rename(
    columns={job_id_col: "job_id", start_col: "start_s", finish_col: "end_s"}
)
lifespan_df["job_key"] = lifespan_df["job_id"].map(canonical_job_id)

lifespan_df = lifespan_df.merge(
    start_events[["job_key", "class", "replica"]],
    on="job_key",
    how="left",
)

print("jobs:", len(jobs))
print("start events:", len(start_events))
print("matched jobs:", lifespan_df["replica"].notna().sum())

if lifespan_df["replica"].notna().sum() == 0:
    print("sample jobs job_id:", jobs["job_id"].head(5).tolist())
    print("sample decisions job_id:", start_events["job_id"].head(5).tolist())
    raise ValueError("No matching job IDs between out__jobs.csv and scheduler_decisions.csv.")

lifespan_df = lifespan_df.dropna(subset=["class", "replica", "start_s", "end_s"]).copy()
lifespan_df["replica"] = lifespan_df["replica"].astype(int)
lifespan_df["replica_label"] = "r" + lifespan_df["replica"].astype(str)
lifespan_df["worker"] = lifespan_df["job_key"]

origin = pd.Timestamp("2000-01-01")
lifespan_df["start_time"] = origin + pd.to_timedelta(lifespan_df["start_s"].astype(float), unit="s")
lifespan_df["end_time"] = origin + pd.to_timedelta(lifespan_df["end_s"].astype(float), unit="s")

if MAX_WORKERS_PER_REPLICA is not None:
    lifespan_df = (
        lifespan_df.sort_values(["replica", "start_s"])
        .groupby("replica", group_keys=False)
        .head(MAX_WORKERS_PER_REPLICA)
    )

replica_order = [f"r{i}" for i in sorted(lifespan_df["replica"].unique())]
print("replicas:", replica_order)

fig_per_replica = px.timeline(
    lifespan_df,
    x_start="start_time",
    x_end="end_time",
    y="worker",
    color="class",
    facet_row="replica_label",
    hover_data={
        "replica_label": True,
        "worker": True,
        "class": True,
        "start_s": ":.3f",
        "end_s": ":.3f",
        "start_time": False,
        "end_time": False,
    },
    title=f"Connection lifespans faceted by replica, {LIFESPAN_POLICY}",
    facet_row_spacing=0.025,
    color_discrete_map=CLASS_COLORS,
    category_orders={"replica_label": replica_order, "class": CLASS_ORDER},
)

fig_per_replica.update_traces(marker=dict(line=dict(width=0.3)))
fig_per_replica.update_layout(
    height=max(1000, 170 * len(replica_order)),
    margin=dict(t=80, b=60, l=80, r=20),
    legend_title_text="Class",
    font=dict(size=10),
)
fig_per_replica.update_yaxes(title_text="Worker/session", autorange="reversed", matches=None)
fig_per_replica.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))

for axis in fig_per_replica.layout:
    if axis.startswith("xaxis"):
        fig_per_replica.layout[axis].update(
            showticklabels=True,
            tickformat="%M:%S",
            tickangle=-45,
            dtick=5000,
        )

fig_per_replica.show()

html_path = OUT / f"{LIFESPAN_POLICY}_connection_lifespans_by_replica.html"
fig_per_replica.write_html(html_path)
print(f"Wrote: {html_path}")
```
