## Metric definitions

For scheduler $p$, priority class $c$, pressure level $T$, and latency quantile $q$:

$$
L_q(p,c,T)=Q_q\left(\{\ell_i \mid class_i=c,\ p_i=p,\ T_i=T\}\right)
$$

where $\ell_i$ is the simulated Q1 latency of query $i$.

Overall Q1 latency pools all queries from the run:

$$
L_q(p,overall,T)=Q_q\left(\{\ell_i \mid p_i=p,\ T_i=T\}\right)
$$

Local Q1 concurrency at start is summarized as:

$$
C_q(p,c,T)=Q_q\left(\{x_i \mid class_i=c,\ p_i=p,\ T_i=T\}\right)
$$

where $x_i$ is the number of already-running Q1 queries on the selected replica when query $i$ starts.

Priority-ordering violations are summarized by:

$$
G_q(p,T)=\max(0,L_q(p,enterprise,T)-L_q(p,premium,T)) +
\max(0,L_q(p,premium,T)-L_q(p,freemium,T))
$$

A value of zero means that the measured latency ordering is consistent with the priority order.

For replica-concentration metrics, let $A_{r,c}$ be the time integral of active Q1 queries of class $c$ on replica $r$. Then:

$$
share_{r,c}=\frac{A_{r,c}}{\sum_j A_{j,c}},\qquad
HHI_c=\sum_r share_{r,c}^2,\qquad
effective\_replicas_c=\frac{1}{HHI_c}
$$

For PLB-to-baseline comparisons:

$$
ratio_q(c,b,T)=\frac{L_q(PLB,c,T)}{L_q(b,c,T)}
$$

Dedicated static splits are plotted as simulated runs. No averaging is applied across different static split names, even when two splits assign the same number of replicas to a class.


## N-class PLB and target-repair variants

This section compares only PLB-family algorithms: the direct N-class PLB extension and the target-repair variants. RR, LL, and static schedulers are excluded from these comparison plots.

- `plb_nclass`: direct N-class extension of PLB.
- `global_target_repair_neutral`: target-repair v1 with `E1 P3 F3 M1`.
- `global_target_repair_target_only`: target-repair v2 with `E1 P3 F3 M1` and target-only neutral repair.
- `global_target_repair_neutral_init_4_2_2`: v1 initialized as `E4 P2 F2 M0`.
- `global_target_repair_target_only_init_4_2_2`: v2 initialized as `E4 P2 F2 M0`.
- `global_target_repair_nomixed`: v3 with no Mixed role, no balance cost, and initial allocation `E4 P3 F1`.
- `global_target_repair_v4`: v4 keeps the same no-Mixed design and initial allocation as v3, but changes only how zero/low-inversion targets are ranked.

For a candidate allocation $k$ and active demand $D_c$, class pressure is

$$
p_c(k)=\frac{D_c}{k_c}.
$$

The desired pressure order is

$$
p_E \le p_P \le p_F.
$$

The pressure-inversion cost is

$$
\mathrm{Inv}(k)=\max(0,p_E-p_P)^2+\max(0,p_P-p_F)^2,
$$

with inactive classes omitted from the active-class ordering used by the implementation.

The v1/v2 controllers choose a target using inversion cost, then balance cost, then distance from the current allocation. v3 removes Mixed and removes balance cost. Its target is selected lexicographically by

$$
\left(\mathrm{Inv}(k),\ d(k,k(t)),\ -k_E,\ -k_P,\ -k_F\right),
$$

where

$$
d(k,k(t))=|k_E-k_E(t)|+|k_P-k_P(t)|+|k_F-k_F(t)|.
$$

The negative replica counts are only a deterministic priority-aware tie-break after inversion cost and distance are equal. v3 moves at most one replica per control opportunity toward the newly computed target. Among classes below target, the highest-priority deficit is served first; among classes above target, the lowest-priority surplus donates first.


For $N$ ordered classes $C_1 > C_2 > \dots > C_N$, v4 selects targets lexicographically by

$$
\left(
\mathrm{Inv}(k),
p_{C_N}(k),
p_{C_{N-1}}(k),
\dots,
p_{C_1}(k),
d(k,k(t)),
\mathrm{tie}(k)
\right).
$$

The first criterion still protects the priority ordering. Among allocations with the same minimum inversion cost, v4 reduces pressure starting from the lowest-priority class upward. Distance from the current allocation is used only after these QoS criteria are tied.

For the current three-class experiment this becomes

$$
\left(
\mathrm{Inv}(k),
p_F(k),
p_P(k),
p_E(k),
d(k,k(t)),
-k_E,-k_P,-k_F
\right).
$$

v3 and v4 are identical in all other tested aspects: no Mixed role, floors of one replica per class, initial allocation `E4 P3 F1`, one replica transfer per control opportunity, the same donor/receiver ordering, and the same routing rule.

### Target-repair controller trace

These plots inspect the controller at one pressure level. The allocation trace shows the role counts after each control opportunity. The inversion trace compares the pressure-inversion cost before and after the executed repair move.

For v3, the target-distance trace also shows how many one-replica transfers separate the current allocation from the target selected at that epoch:

$$
\mathrm{moves\_required}=\frac{1}{2}\lVert k^*-k(t)\rVert_1.
$$

This metric is useful for assessing the one-move-per-epoch actuation rule. If targets are frequently several moves away, especially for larger $K$, the actuation rule may become the response-time bottleneck even when target selection is correct.

The controller inversion cost is not the same as the measured latency priority gap. The controller uses active demand and role allocation; the latency priority gap is computed later from simulated Q1 latency.


For controller debugging, the notebook focuses the detailed traces on v3 and v4 and restricts them to the active workload window. It also reports allocation residency: the number of distinct allocations visited and the fraction of control epochs spent in the dominant allocation. This directly measures whether a controller is effectively stuck in one allocation.


```python
CONTROL_TRACE_END_S = 180.0
CONTROLLER_FOCUS_SCHEDULERS = [
    scheduler
    for scheduler in [
        "global_target_repair_nomixed",
        "global_target_repair_v4",
    ]
    if scheduler in ALGO_AVAILABLE
]

def read_target_repair_decisions(
    scheduler,
    T=600,
    ratio=SELECTED_RATIO,
    seed=42,
    end_s=CONTROL_TRACE_END_S,
):
    path = (
        CAMPAIGN_DIR
        / "K8"
        / f"T{T}"
        / f"ratio_{ratio}"
        / f"seed{seed}"
        / scheduler
        / "target_repair_decisions.csv"
    )
    if not path.exists():
        print("Missing target-repair decisions:", path)
        return pd.DataFrame()

    decisions = pd.read_csv(path)
    if end_s is not None and "time" in decisions.columns:
        decisions = decisions[decisions["time"] <= float(end_s)].copy()
    return decisions


def plot_target_repair_allocation_trace(scheduler, T=600, ratio=SELECTED_RATIO, seed=42):
    decisions = read_target_repair_decisions(scheduler=scheduler, T=T, ratio=ratio, seed=seed)
    if decisions.empty:
        return

    role_cols = {
        "Enterprise": "after_E",
        "Premium": "after_P",
        "Freemium": "after_F",
        "Mixed": "after_M",
    }
    role_cols = {label: col for label, col in role_cols.items() if col in decisions.columns}

    fig, ax = plt.subplots(figsize=DOUBLE_COL_TALL)
    role_style = {
        "Enterprise": {"color": PALETTE["enterprise"], "linewidth": 1.8},
        "Premium": {"color": PALETTE["premium"], "linewidth": 1.8},
        "Freemium": {"color": PALETTE["freemium"], "linewidth": 1.8},
        "Mixed": {"color": PALETTE["dark_gray"], "linewidth": 1.2, "linestyle": "--"},
    }
    configure_marker_styles(role_style)

    for label, col in role_cols.items():
        ax.step(
            decisions["time"],
            decisions[col],
            where="post",
            label=label,
            **role_style[label],
        )

    ax.set_xlabel("Simulation time (s)")
    ax.set_ylabel("Replicas by role")
    ax.set_title(f"Role allocation trace, {ALGO_LABELS.get(scheduler, scheduler)}, T={T}")
    ax.set_ylim(-0.1, 8.1)
    ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1.0), frameon=False)
    save_plb_figure(fig, f"target_repair_allocation_trace_{scheduler}_T{T}_{ratio}")
    plt.show()


def plot_target_repair_inversion_trace(scheduler, T=600, ratio=SELECTED_RATIO, seed=42):
    decisions = read_target_repair_decisions(scheduler=scheduler, T=T, ratio=ratio, seed=seed)
    if decisions.empty:
        return

    fig, ax = plt.subplots(figsize=DOUBLE_COL_TALL)
    ax.plot(
        decisions["time"],
        decisions["current_inv"],
        label="Before repair",
        color=PALETTE["dark_gray"],
        linewidth=1.2,
        linestyle="--",
    )
    ax.plot(
        decisions["time"],
        decisions["after_inv"],
        label="After repair",
        color=PALETTE["black"],
        linewidth=1.8,
    )
    ax.axhline(0.0, color=PALETTE["dark_gray"], linewidth=0.7, alpha=0.7)
    ax.set_xlabel("Simulation time (s)")
    ax.set_ylabel("Priority-inversion cost")
    ax.set_title(f"Pressure-order repair trace, {ALGO_LABELS.get(scheduler, scheduler)}, T={T}")
    ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1.0), frameon=False)
    save_plb_figure(fig, f"target_repair_inversion_trace_{scheduler}_T{T}_{ratio}")
    plt.show()


def plot_target_repair_distance_trace(scheduler, T=600, ratio=SELECTED_RATIO, seed=42):
    decisions = read_target_repair_decisions(scheduler=scheduler, T=T, ratio=ratio, seed=seed)
    if decisions.empty or "target_moves_required" not in decisions.columns:
        return

    fig, ax = plt.subplots(figsize=DOUBLE_COL_TALL)
    ax.step(
        decisions["time"],
        decisions["target_moves_required"],
        where="post",
        label="Before actuation",
        linewidth=1.8,
    )
    if "after_l1_distance" in decisions.columns:
        ax.step(
            decisions["time"],
            decisions["after_l1_distance"] / 2.0,
            where="post",
            label="After actuation",
            linewidth=1.2,
            linestyle="--",
        )
    ax.axhline(0.0, linewidth=0.7, alpha=0.7)
    ax.set_xlabel("Simulation time (s)")
    ax.set_ylabel("Replica transfers to current target")
    ax.set_title(f"Target distance, {ALGO_LABELS.get(scheduler, scheduler)}, T={T}")
    ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1.0), frameon=False)
    save_plb_figure(fig, f"target_repair_distance_trace_{scheduler}_T{T}_{ratio}")
    plt.show()


def target_repair_control_summary(T=600, ratio=SELECTED_RATIO, seed=42):
    rows = []
    for scheduler in ALGO_TARGET_REPAIR:
        decisions = read_target_repair_decisions(scheduler=scheduler, T=T, ratio=ratio, seed=seed)
        if decisions.empty:
            continue

        move_mask = decisions["move_from"].fillna("") != ""
        row = {
            "scheduler": ALGO_LABELS.get(scheduler, scheduler),
            "control_epochs": len(decisions),
            "role_moves": int(move_mask.sum()),
            "move_fraction": float(move_mask.mean()),
            "fraction_current_inv_positive": float((decisions["current_inv"] > 0).mean()),
            "fraction_after_inv_positive": float((decisions["after_inv"] > 0).mean()),
            "mean_inv_reduction": float((decisions["current_inv"] - decisions["after_inv"]).mean()),
            "neutral_moves": int(decisions["reason"].astype(str).str.startswith("neutral_").sum()),
        }

        if "target_moves_required" in decisions.columns:
            row["mean_target_moves_required"] = float(decisions["target_moves_required"].mean())
            row["max_target_moves_required"] = int(decisions["target_moves_required"].max())
        else:
            row["mean_target_moves_required"] = np.nan
            row["max_target_moves_required"] = np.nan

        if "after_M" in decisions.columns:
            row["fraction_after_M_positive"] = float((decisions["after_M"] > 0).mean())
            row["mean_after_M"] = float(decisions["after_M"].mean())

            q_path = (
                CAMPAIGN_DIR
                / "K8"
                / f"T{T}"
                / f"ratio_{ratio}"
                / f"seed{seed}"
                / scheduler
                / "query_events.csv"
            )
            mixed_starts = np.nan
            if q_path.exists():
                queries = pd.read_csv(q_path)
                if "replica_role_at_start" in queries.columns:
                    mixed_starts = int((queries["replica_role_at_start"] == "mixed").sum())
            row["mixed_query_starts"] = mixed_starts
        else:
            row["fraction_after_M_positive"] = np.nan
            row["mean_after_M"] = np.nan
            row["mixed_query_starts"] = np.nan

        rows.append(row)

    if rows:
        display(pd.DataFrame(rows))


def allocation_residency_summary(T=600, ratio=SELECTED_RATIO, seed=42):
    rows = []
    for scheduler in CONTROLLER_FOCUS_SCHEDULERS:
        decisions = read_target_repair_decisions(
            scheduler=scheduler,
            T=T,
            ratio=ratio,
            seed=seed,
        )
        if decisions.empty:
            continue

        alloc_cols = [col for col in ["after_E", "after_P", "after_F", "after_M"] if col in decisions.columns]
        allocation_labels = decisions[alloc_cols].astype(int).astype(str).agg("-".join, axis=1)
        counts = allocation_labels.value_counts()
        move_mask = decisions["move_from"].fillna("") != ""

        rows.append({
            "scheduler": ALGO_LABELS.get(scheduler, scheduler),
            "control_epochs": len(decisions),
            "role_moves": int(move_mask.sum()),
            "distinct_allocations": int(counts.size),
            "dominant_allocation": counts.index[0],
            "dominant_allocation_fraction": float(counts.iloc[0] / counts.sum()),
        })

    if rows:
        display(pd.DataFrame(rows))


for scheduler in CONTROLLER_FOCUS_SCHEDULERS:
    plot_target_repair_allocation_trace(scheduler=scheduler, T=600)
    plot_target_repair_inversion_trace(scheduler=scheduler, T=600)
    plot_target_repair_distance_trace(scheduler=scheduler, T=600)

allocation_residency_summary(T=600)
target_repair_control_summary(T=600)
```
